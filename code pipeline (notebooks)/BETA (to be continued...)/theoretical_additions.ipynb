# Bonus Theoretical Additions

This notebook verifies the Information Theory implementations:
1. **Mutual Information (Feature Selection)**
2. **Huffman Compression Ratio (Benign vs. Attack)**

```python
import sys
import os
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

# Ensure the local modules can be imported
if os.path.abspath('.') not in sys.path:
    sys.path.append(os.path.abspath('.'))

from theory_metrics import calculate_mutual_information
from huffman_eval import evaluate_huffman_compression

DATASET_PATH = "../data/processed.parquet"
```

## 1. Mutual Information (Addition A)
Calculating $I(X; Y)$ to mathematically prove feature importance.

```python
mi_df = calculate_mutual_information(DATASET_PATH)
display(mi_df)

# Plotting
plt.figure(figsize=(10, 5))
plt.bar(mi_df['Feature'], mi_df['Mutual Information (bits)'], color='teal')
plt.xticks(rotation=45, ha='right')
plt.title('Mutual Information of Features w.r.t Label')
plt.ylabel('I(X; Y) in bits')
plt.tight_layout()
plt.show()
```

## 2. Huffman Compression Ratio (Addition C)
Comparing compression on a 5-minute benign window vs a 5-minute attack window using `Tot_Fwd_Pkts_sym`.

```python
huffman_df = evaluate_huffman_compression(DATASET_PATH)
display(huffman_df)

# Plotting Compression Ratio
plt.figure(figsize=(8, 4))
plt.bar(huffman_df['Traffic Type'], huffman_df['Compression Ratio'], color=['blue', 'red'])
plt.title('Huffman Compression Ratio (Higher is better)')
plt.ylabel('Ratio (Original / Compressed)')
plt.tight_layout()
plt.show()
```
